# 1D CNN Autoencoder (strict normal-only)

Sequence-to-sequence 1D CNN encoder-decoder that reconstructs each 30-step motor window.

The model is trained strictly on normal-running sessions only. Motor-off, transition, and fault files are used only for evaluation.

The notebook includes file-wise cross-validation, final evaluation, manual testing, Keras/TFLite/ONNX export, and a complete direct C header for STM32 deployment.

```python
# ==========================================================
# CELL 1 — imports
# ==========================================================
import os
import glob
import random
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    Conv1D,
    MaxPooling1D,
    UpSampling1D,
)
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
)
```

```python
# ==========================================================
# CELL 2 — configuration
# ==========================================================
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Change only this path.
BASE_PATH = (
    r"../../data"
    r"\dataset\Motor_data_model"
)

# Raw data values are stored as scaled integers.
SCALE_DIVISOR = 1_000_000.0

FEATURE_COLUMNS = [
    "T_slope_10s",
    "PWM",
    "Irms",
    "Vrms",
    "Current_Variance",
    "Current_Crest_Factor",
    "Vib_RMS",
    "Vib_Min",
    "Vib_Q1",
    "Vib_Median",
    "Vib_Q3",
    "Vib_Max",
    "Vib_Variance",
    "Vib_Skewness",
]

N_FEATURES = len(FEATURE_COLUMNS)

WINDOW_LEN = 30
WINDOW_STEP = 10
ROW_PERIOD_SEC = 0.015

N_SPLITS = 5
EPOCHS = 100
BATCH_SIZE = 64

RECON_THRESHOLD_PERCENTILE = 99.0
FEATURE_THRESHOLD_PERCENTILE = 99.0

# Available modes:
# "mean"   = reconstruction MSE only
# "max"    = maximum feature-ratio only
# "hybrid" = either condition can trigger anomaly
SCORE_MODE = "hybrid"

CLIP_SCALED_INPUT = False

FILE_ANOMALY_FRACTION_THRESHOLD = 0.10

OUTPUT_DIR = "1dcnn_ae_output"

CV_DIR = os.path.join(
    OUTPUT_DIR,
    "cross_validation",
)

FINAL_DIR = os.path.join(
    OUTPUT_DIR,
    "final_model",
)

FINAL_EVAL_DIR = os.path.join(
    OUTPUT_DIR,
    "final_model_evaluation",
)

MANUAL_TEST_DIR = os.path.join(
    OUTPUT_DIR,
    "manual_test_results",
)

for folder in [
    CV_DIR,
    FINAL_DIR,
    FINAL_EVAL_DIR,
    MANUAL_TEST_DIR,
]:
    os.makedirs(
        folder,
        exist_ok=True,
    )

print("TensorFlow:", tf.__version__)
print("Features:", N_FEATURES)
print("Input shape:", (WINDOW_LEN, N_FEATURES))
```

```python
# ==========================================================
# CELL 3 — normal and anomaly file detection
# ==========================================================
def normalized_path(path):
    return str(path).replace("\\", "/").lower()


def normal_state_for(path):
    p = normalized_path(path)
    stem = Path(p).stem.lower()

    excluded_words = [
        "motor_off",
        "motor_being_on",
        "motor_being_off",
        "transition",
        "off_on_spikes",
    ]

    if any(word in p for word in excluded_words):
        return None

    if (
        "/normal/" in p
        or stem.startswith("normal_")
        or stem == "normal"
    ):
        return "normal_running"

    return None


def anomaly_label_for(path):
    p = normalized_path(path)

    if "motor_being_on" in p:
        return "motor_being_on"

    if "motor_being_off" in p:
        return "motor_being_off"

    if (
        "/motor_off/" in p
        or "motor_off" in Path(p).stem.lower()
    ):
        return "motor_off"

    if (
        "electrical fault" in p
        or "abnormal_current" in p
    ):
        return "electrical_fault"

    if (
        "mechanical fault" in p
        or "abnormal_vibration" in p
    ):
        return "mechanical_fault"

    if (
        "thermal fault" in p
        or "abnormal_temp" in p
    ):
        return "thermal_fault"

    return None
```

```python
# ==========================================================
# CELL 4 — discover dataset files
# ==========================================================
def discover_files(base_path):
    files = glob.glob(
        os.path.join(
            base_path,
            "**",
            "*.xlsx",
        ),
        recursive=True,
    )

    files += glob.glob(
        os.path.join(
            base_path,
            "**",
            "*.csv",
        ),
        recursive=True,
    )

    return sorted(set(files))


files = discover_files(BASE_PATH)

if not files:
    raise RuntimeError(
        f"No CSV or Excel files found under: {BASE_PATH}"
    )

print(
    "Total files discovered:",
    len(files),
)
```

```python
# ==========================================================
# CELL 5 — read one file
# ==========================================================
def read_one(path):
    path = Path(path)

    if path.suffix.lower() == ".xlsx":
        df = pd.read_excel(path)

    elif path.suffix.lower() == ".csv":
        df = pd.read_csv(path)

    else:
        raise ValueError(
            f"Unsupported file type: {path}"
        )

    missing = [
        column
        for column in FEATURE_COLUMNS
        if column not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing columns in {path.name}: {missing}"
        )

    clean = df[FEATURE_COLUMNS].apply(
        pd.to_numeric,
        errors="coerce",
    )

    clean = clean / SCALE_DIVISOR

    clean = clean.replace(
        [np.inf, -np.inf],
        np.nan,
    )

    clean = clean.dropna().reset_index(drop=True)

    if len(clean) < WINDOW_LEN:
        raise ValueError(
            f"{path.name} has only {len(clean)} valid rows. "
            f"At least {WINDOW_LEN} rows are required."
        )

    return clean
```

```python
# ==========================================================
# CELL 6 — load only normal WITH-LOAD and all anomaly sessions
# ==========================================================
def is_normal_with_load(path):
    """
    Return True only for normal files containing motor data
    recorded WITH load.
    """
    p = normalized_path(path)

    # Make different naming styles comparable.
    p = (
        p.replace(" ", "_")
         .replace("-", "_")
    )

    # Check negative markers first.
    without_load_markers = [
        "without_load",
        "withoutload",
        "no_load",
        "noload",
        "unloaded",
    ]

    if any(marker in p for marker in without_load_markers):
        return False

    with_load_markers = [
        "with_load",
        "withload",
        "normal_load",
        "normal_loaded",
    ]

    return any(marker in p for marker in with_load_markers)


healthy_sessions = []
fault_sessions = []
excluded_normal_files = []

for file_path in files:
    normal_state = normal_state_for(file_path)
    fault_label = anomaly_label_for(file_path)

    if normal_state is None and fault_label is None:
        print(
            "SKIP unrecognized:",
            os.path.relpath(file_path, BASE_PATH),
        )
        continue

    # Exclude normal files recorded without load.
    # Fault files are not affected by this condition.
    if (
        normal_state is not None
        and not is_normal_with_load(file_path)
    ):
        relative_path = os.path.relpath(
            file_path,
            BASE_PATH,
        )

        excluded_normal_files.append(
            relative_path
        )

        print(
            "EXCLUDED normal without load:",
            relative_path,
        )
        continue

    try:
        df = read_one(file_path)

    except Exception as error:
        print(
            "SKIP read error:",
            os.path.relpath(
                file_path,
                BASE_PATH,
            ),
            "|",
            error,
        )
        continue

    record = {
        "X": df.to_numpy(
            dtype=np.float32
        ),

        "file": os.path.relpath(
            file_path,
            BASE_PATH,
        ),

        "rows": len(df),
    }

    if normal_state is not None:
        record.update({
            "state": normal_state,
            "load_condition": "with_load",
            "binary_label": 0,
        })

        healthy_sessions.append(
            record
        )

        print(
            f"N | {'normal_with_load':20s} | "
            f"{len(df):6d} | "
            f"{record['file']}"
        )

    else:
        record.update({
            "fault_label": fault_label,
            "binary_label": 1,
        })

        fault_sessions.append(
            record
        )

        print(
            f"A | {fault_label:20s} | "
            f"{len(df):6d} | "
            f"{record['file']}"
        )


if not healthy_sessions:
    raise RuntimeError(
        "No normal WITH-LOAD files were loaded. "
        "Check the normal filenames or folder names."
    )

if len(healthy_sessions) < N_SPLITS:
    raise RuntimeError(
        f"Only {len(healthy_sessions)} normal with-load files "
        f"were loaded, but N_SPLITS={N_SPLITS}."
    )

if not fault_sessions:
    raise RuntimeError(
        "No anomaly files were loaded."
    )


print("\nDATASET SUMMARY")

print(
    "Normal WITH-LOAD files used:",
    len(healthy_sessions),
)

print(
    "Normal WITHOUT-LOAD files excluded:",
    len(excluded_normal_files),
)

print(
    "Fault files:",
    len(fault_sessions),
)


print("\nNORMAL WITH-LOAD FILES USED")

for session in healthy_sessions:
    print(
        "  USED:",
        session["file"],
    )


print("\nNORMAL WITHOUT-LOAD FILES EXCLUDED")

for file_name in excluded_normal_files:
    print(
        "  EXCLUDED:",
        file_name,
    )


print("\nFAULT FILE DISTRIBUTION")

print(
    pd.Series([
        session["fault_label"]
        for session in fault_sessions
    ]).value_counts()
)
```

```python
# ==========================================================
# CELL 7 — window generation
# ==========================================================
def make_windows_from_array(
    X_raw,
    scaler,
):
    X_scaled = scaler.transform(
        X_raw
    ).astype(np.float32)

    if CLIP_SCALED_INPUT:
        X_scaled = np.clip(
            X_scaled,
            0.0,
            1.0,
        )

    starts = np.arange(
        0,
        len(X_scaled) - WINDOW_LEN + 1,
        WINDOW_STEP,
        dtype=np.int64,
    )

    if len(starts) == 0:
        return None, None

    windows = np.stack([
        X_scaled[
            start:start + WINDOW_LEN
        ]
        for start in starts
    ]).astype(np.float32)

    return windows, starts


def build_windows_from_sessions(
    sessions,
    scaler,
):
    all_windows = []

    for session in sessions:
        windows, _ = make_windows_from_array(
            session["X"],
            scaler,
        )

        if windows is not None:
            all_windows.append(
                windows
            )

    if not all_windows:
        return np.empty(
            (
                0,
                WINDOW_LEN,
                N_FEATURES,
            ),
            dtype=np.float32,
        )

    return np.concatenate(
        all_windows,
        axis=0,
    ).astype(np.float32)
```

```python
# ==========================================================
# CELL 8 — 1D CNN autoencoder model
# ==========================================================
def window_mse_loss(
    y_true,
    y_pred,
):
    return tf.reduce_mean(
        tf.square(
            y_true - y_pred
        ),
        axis=[1, 2],
    )


def build_autoencoder_model(
    window_len,
    n_features,
):
    inputs = Input(
        shape=(
            window_len,
            n_features,
        ),
        name="float_input",
    )

    x = Conv1D(
        filters=16,
        kernel_size=3,
        padding="same",
        activation="relu",
        name="encoder_conv",
    )(inputs)

    x = MaxPooling1D(
        pool_size=2,
        padding="valid",
        name="encoder_pool",
    )(x)

    latent = Conv1D(
        filters=8,
        kernel_size=3,
        padding="same",
        activation="relu",
        name="latent_conv",
    )(x)

    x = UpSampling1D(
        size=2,
        name="decoder_upsample",
    )(latent)

    x = Conv1D(
        filters=16,
        kernel_size=3,
        padding="same",
        activation="relu",
        name="decoder_conv",
    )(x)

    outputs = Conv1D(
        filters=n_features,
        kernel_size=3,
        padding="same",
        activation="linear",
        name="reconstruction_output",
    )(x)

    model = Model(
        inputs=inputs,
        outputs=outputs,
        name="pure_normal_1dcnn_autoencoder",
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=1e-3
        ),
        loss=window_mse_loss,
    )

    return model


def make_callbacks():
    return [
        EarlyStopping(
            monitor="val_loss",
            mode="min",
            patience=15,
            restore_best_weights=True,
        ),

        ReduceLROnPlateau(
            monitor="val_loss",
            mode="min",
            factor=0.5,
            patience=5,
            min_lr=1e-5,
        ),
    ]


temporary_model = build_autoencoder_model(
    WINDOW_LEN,
    N_FEATURES,
)

temporary_model.summary()

del temporary_model
```

```python
# ==========================================================
# CELL 9 — anomaly scoring and threshold calibration
# ==========================================================
def anomaly_score(
    model,
    X,
):
    reconstruction = model.predict(
        X,
        verbose=0,
    )

    per_feature_error = np.mean(
        np.square(
            X - reconstruction
        ),
        axis=1,
    ).astype(np.float32)

    reconstruction_mse = np.mean(
        per_feature_error,
        axis=1,
    ).astype(np.float32)

    worst_feature_ids = np.argmax(
        per_feature_error,
        axis=1,
    )

    return (
        reconstruction_mse,
        per_feature_error,
        worst_feature_ids,
    )


def calibrate_thresholds(
    model,
    X_normal,
):
    (
        reconstruction_mse,
        per_feature_error,
        _,
    ) = anomaly_score(
        model,
        X_normal,
    )

    reconstruction_threshold = float(
        np.percentile(
            reconstruction_mse,
            RECON_THRESHOLD_PERCENTILE,
        )
    )

    feature_thresholds = np.percentile(
        per_feature_error,
        FEATURE_THRESHOLD_PERCENTILE,
        axis=0,
    ).astype(np.float32)

    feature_ratios = (
        per_feature_error
        / np.maximum(
            feature_thresholds.reshape(
                1,
                -1,
            ),
            1e-12,
        )
    )

    maximum_feature_ratio = np.max(
        feature_ratios,
        axis=1,
    )

    max_ratio_threshold = float(
        np.percentile(
            maximum_feature_ratio,
            FEATURE_THRESHOLD_PERCENTILE,
        )
    )

    return (
        reconstruction_threshold,
        feature_thresholds,
        max_ratio_threshold,
    )


def decisions_from_errors(
    reconstruction_mse,
    per_feature_error,
    reconstruction_threshold,
    feature_thresholds,
    max_ratio_threshold,
):
    mean_flag = (
        reconstruction_mse
        > reconstruction_threshold
    )

    feature_ratios = (
        per_feature_error
        / np.maximum(
            feature_thresholds.reshape(
                1,
                -1,
            ),
            1e-12,
        )
    )

    maximum_feature_ratio = np.max(
        feature_ratios,
        axis=1,
    )

    feature_flag = (
        maximum_feature_ratio
        > max_ratio_threshold
    )

    if SCORE_MODE == "mean":
        flags = mean_flag

    elif SCORE_MODE == "max":
        flags = feature_flag

    elif SCORE_MODE == "hybrid":
        flags = (
            mean_flag
            | feature_flag
        )

    else:
        raise ValueError(
            "SCORE_MODE must be mean, max, or hybrid."
        )

    return (
        flags.astype(np.int64),
        maximum_feature_ratio.astype(
            np.float32
        ),
    )
```

```python
# ==========================================================
# CELL 10 — file-level and window-level predictions
# ==========================================================
def aggregate_predictions(
    sessions,
    model,
    scaler,
    reconstruction_threshold,
    feature_thresholds,
    max_ratio_threshold,
):
    file_rows = []
    window_rows = []

    for session in sessions:
        windows, starts = make_windows_from_array(
            session["X"],
            scaler,
        )

        if windows is None:
            continue

        (
            reconstruction_mse,
            per_feature_error,
            worst_feature_ids,
        ) = anomaly_score(
            model,
            windows,
        )

        (
            flags,
            maximum_feature_ratio,
        ) = decisions_from_errors(
            reconstruction_mse,
            per_feature_error,
            reconstruction_threshold,
            feature_thresholds,
            max_ratio_threshold,
        )

        true_label = int(
            session["binary_label"]
        )

        state = session.get(
            "state",
            session.get(
                "fault_label",
                "unknown",
            ),
        )

        anomaly_fraction = float(
            np.mean(flags)
        )

        file_prediction = int(
            anomaly_fraction
            >= FILE_ANOMALY_FRACTION_THRESHOLD
        )

        file_rows.append({
            "file": session["file"],
            "state": state,
            "true_label": true_label,
            "pred_label": file_prediction,
            "windows": len(windows),
            "anomaly_windows": int(
                np.sum(flags)
            ),
            "anomaly_fraction":
                anomaly_fraction,
            "mean_mse": float(
                np.mean(
                    reconstruction_mse
                )
            ),
            "max_mse": float(
                np.max(
                    reconstruction_mse
                )
            ),
            "reconstruction_threshold":
                reconstruction_threshold,
            "max_ratio_threshold":
                max_ratio_threshold,
        })

        for window_id in range(
            len(windows)
        ):
            window_rows.append({
                "file":
                    session["file"],

                "state":
                    state,

                "window":
                    window_id,

                "start_row":
                    int(
                        starts[window_id]
                    ),

                "true_label":
                    true_label,

                "pred_label":
                    int(
                        flags[window_id]
                    ),

                "reconstruction_mse":
                    float(
                        reconstruction_mse[
                            window_id
                        ]
                    ),

                "reconstruction_threshold":
                    reconstruction_threshold,

                "max_feature_ratio":
                    float(
                        maximum_feature_ratio[
                            window_id
                        ]
                    ),

                "max_ratio_threshold":
                    max_ratio_threshold,

                "worst_feature":
                    FEATURE_COLUMNS[
                        int(
                            worst_feature_ids[
                                window_id
                            ]
                        )
                    ],
            })

    return (
        pd.DataFrame(
            file_rows
        ),
        pd.DataFrame(
            window_rows
        ),
    )
```

```python
# ==========================================================
# CELL 11 — build cross-validation table
# ==========================================================
all_sessions = (
    healthy_sessions
    + fault_sessions
)

cv_table = pd.DataFrame({
    "index": np.arange(
        len(all_sessions)
    ),

    "label": [
        session["binary_label"]
        for session in all_sessions
    ],

    "group": [
        session["file"]
        for session in all_sessions
    ],
})

print(
    cv_table["label"].value_counts()
)
```

```python
# ==========================================================
# CELL 12 — five-fold session-wise cross-validation
# ==========================================================
splitter = StratifiedGroupKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=SEED,
)

cv_metrics = []
cv_file_results = []
cv_window_results = []
cv_test_files = []

for fold, (
    train_index,
    test_index,
) in enumerate(
    splitter.split(
        cv_table,
        cv_table["label"],
        groups=cv_table["group"],
    ),
    start=1,
):
    print(
        "\n"
        + "=" * 70
    )

    print(
        f"CV FOLD {fold}/{N_SPLITS}"
    )

    print(
        "=" * 70
    )

    train_sessions = [
        all_sessions[index]
        for index in train_index
    ]

    test_sessions = [
        all_sessions[index]
        for index in test_index
    ]

    print("\nTEST FILES")

    for session in test_sessions:
        state = session.get(
            "state",
            session.get(
                "fault_label",
                "unknown",
            ),
        )

        cv_test_files.append({
            "fold": fold,
            "file": session["file"],
            "state": state,
            "true_label":
                session["binary_label"],
        })

        print(
            f"label={session['binary_label']} | "
            f"state={state} | "
            f"file={session['file']}"
        )

    normal_train_sessions = [
        session
        for session in train_sessions
        if session["binary_label"] == 0
    ]

    rng = np.random.default_rng(
        SEED + fold
    )

    shuffled_ids = rng.permutation(
        len(normal_train_sessions)
    )

    number_calibration_files = max(
        1,
        round(
            0.20
            * len(shuffled_ids)
        ),
    )

    calibration_ids = set(
        shuffled_ids[
            :number_calibration_files
        ].tolist()
    )

    fit_sessions = [
        session
        for index, session
        in enumerate(
            normal_train_sessions
        )
        if index not in calibration_ids
    ]

    calibration_sessions = [
        session
        for index, session
        in enumerate(
            normal_train_sessions
        )
        if index in calibration_ids
    ]

    print(
        "\nNormal fit files:",
        len(fit_sessions),
    )

    print(
        "Normal calibration files:",
        len(calibration_sessions),
    )

    scaler = MinMaxScaler(
        feature_range=(0, 1)
    )

    scaler.fit(
        np.concatenate([
            session["X"]
            for session in fit_sessions
        ])
    )

    X_fit = build_windows_from_sessions(
        fit_sessions,
        scaler,
    )

    X_calibration = (
        build_windows_from_sessions(
            calibration_sessions,
            scaler,
        )
    )

    tf.keras.backend.clear_session()

    model = build_autoencoder_model(
        WINDOW_LEN,
        N_FEATURES,
    )

    model.fit(
        X_fit,
        X_fit,

        validation_data=(
            X_calibration,
            X_calibration,
        ),

        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=make_callbacks(),
        verbose=1,
        shuffle=True,
    )

    (
        reconstruction_threshold,
        feature_thresholds,
        max_ratio_threshold,
    ) = calibrate_thresholds(
        model,
        X_calibration,
    )

    (
        fold_file_df,
        fold_window_df,
    ) = aggregate_predictions(
        test_sessions,
        model,
        scaler,
        reconstruction_threshold,
        feature_thresholds,
        max_ratio_threshold,
    )

    fold_file_df.insert(
        0,
        "fold",
        fold,
    )

    fold_window_df.insert(
        0,
        "fold",
        fold,
    )

    cv_file_results.append(
        fold_file_df
    )

    cv_window_results.append(
        fold_window_df
    )

    y_true = fold_window_df[
        "true_label"
    ].to_numpy()

    y_pred = fold_window_df[
        "pred_label"
    ].to_numpy()

    y_score = fold_window_df[
        "reconstruction_mse"
    ].to_numpy()

    fold_result = {
        "fold": fold,

        "accuracy": accuracy_score(
            y_true,
            y_pred,
        ),

        "balanced_accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred,
            ),

        "precision": precision_score(
            y_true,
            y_pred,
            zero_division=0,
        ),

        "recall": recall_score(
            y_true,
            y_pred,
            zero_division=0,
        ),

        "f1": f1_score(
            y_true,
            y_pred,
            zero_division=0,
        ),

        "roc_auc": (
            roc_auc_score(
                y_true,
                y_score,
            )
            if len(
                np.unique(y_true)
            ) == 2
            else np.nan
        ),

        "reconstruction_threshold":
            reconstruction_threshold,

        "max_ratio_threshold":
            max_ratio_threshold,
    }

    cv_metrics.append(
        fold_result
    )

    print(
        "\nWINDOW-LEVEL FOLD RESULTS"
    )

    print(
        pd.Series(
            fold_result
        )
    )

    print(
        classification_report(
            y_true,
            y_pred,
            target_names=[
                "normal",
                "anomaly",
            ],
            zero_division=0,
        )
    )
```

```python
# ==========================================================
# CELL 17.1 — export complete 1D CNN model as C header
# ==========================================================
C_HEADER_PATH = os.path.join(
    FINAL_DIR,
    "1dcnn_autoencoder_model.h",
)

K1, B1 = final_model.get_layer(
    "encoder_conv"
).get_weights()

K2, B2 = final_model.get_layer(
    "latent_conv"
).get_weights()

K3, B3 = final_model.get_layer(
    "decoder_conv"
).get_weights()

K4, B4 = final_model.get_layer(
    "reconstruction_output"
).get_weights()


def write_float_array(
    file,
    name,
    values,
):
    values = np.asarray(
        values,
        dtype=np.float32,
    ).reshape(-1)

    file.write(
        f"static const float "
        f"{name}[{len(values)}] = {{\n"
    )

    for start in range(
        0,
        len(values),
        8,
    ):
        line = ", ".join(
            f"{float(value):.9g}f"
            for value in values[
                start:start + 8
            ]
        )

        file.write(
            f"    {line},\n"
        )

    file.write(
        "};\n\n"
    )


with open(
    C_HEADER_PATH,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "#ifndef CNN_AE_MODEL_H\n"
    )
    file.write(
        "#define CNN_AE_MODEL_H\n\n"
    )
    file.write(
        "#include <math.h>\n\n"
    )

    file.write(
        f"#define CNN_NF {N_FEATURES}\n"
    )
    file.write(
        f"#define CNN_WL {WINDOW_LEN}\n"
    )
    file.write(
        "#define CNN_POOL_LEN 15\n"
    )
    file.write(
        "#define CNN_F1 16\n"
    )
    file.write(
        "#define CNN_F2 8\n"
    )
    file.write(
        f"#define CNN_DIV "
        f"{SCALE_DIVISOR:.9g}f\n"
    )
    file.write(
        f"#define CNN_RT "
        f"{FINAL_RECON_THRESHOLD:.9g}f\n"
    )
    file.write(
        f"#define CNN_MRT "
        f"{FINAL_MAX_RATIO_THRESHOLD:.9g}f\n\n"
    )

    write_float_array(
        file,
        "CNN_MIN",
        final_scaler.data_min_,
    )

    write_float_array(
        file,
        "CNN_MAX",
        final_scaler.data_max_,
    )

    write_float_array(
        file,
        "CNN_FT",
        FINAL_FEATURE_THRESHOLDS,
    )

    write_float_array(
        file,
        "CNN_K1",
        K1,
    )

    write_float_array(
        file,
        "CNN_B1",
        B1,
    )

    write_float_array(
        file,
        "CNN_K2",
        K2,
    )

    write_float_array(
        file,
        "CNN_B2",
        B2,
    )

    write_float_array(
        file,
        "CNN_K3",
        K3,
    )

    write_float_array(
        file,
        "CNN_B3",
        B3,
    )

    write_float_array(
        file,
        "CNN_K4",
        K4,
    )

    write_float_array(
        file,
        "CNN_B4",
        B4,
    )

    file.write(r'''
static inline float cnn_relu(float x)
{
    return x > 0.0f ? x : 0.0f;
}


static inline void cnn_conv_same(
    const float *input,
    int length,
    int input_channels,
    const float *kernel,
    const float *bias,
    int output_channels,
    int use_relu,
    float *output)
{
    for (int t = 0; t < length; t++)
    {
        for (int oc = 0; oc < output_channels; oc++)
        {
            float sum = bias[oc];

            for (int k = 0; k < 3; k++)
            {
                int source_t = t + k - 1;

                if (
                    source_t < 0
                    || source_t >= length
                )
                {
                    continue;
                }

                for (int ic = 0; ic < input_channels; ic++)
                {
                    int input_index =
                        source_t * input_channels + ic;

                    int kernel_index =
                        (
                            k * input_channels + ic
                        )
                        * output_channels + oc;

                    sum +=
                        input[input_index]
                        * kernel[kernel_index];
                }
            }

            if (use_relu)
            {
                sum = cnn_relu(sum);
            }

            output[
                t * output_channels + oc
            ] = sum;
        }
    }
}


static inline void cnn_pool_2(
    const float *input,
    int channels,
    float *output)
{
    for (int t = 0; t < CNN_POOL_LEN; t++)
    {
        for (int c = 0; c < channels; c++)
        {
            float a =
                input[(2 * t) * channels + c];

            float b =
                input[(2 * t + 1) * channels + c];

            output[t * channels + c] =
                fmaxf(a, b);
        }
    }
}


static inline void cnn_upsample_2(
    const float *input,
    int channels,
    float *output)
{
    for (int t = 0; t < CNN_POOL_LEN; t++)
    {
        for (int c = 0; c < channels; c++)
        {
            float value =
                input[t * channels + c];

            output[
                (2 * t) * channels + c
            ] = value;

            output[
                (2 * t + 1) * channels + c
            ] = value;
        }
    }
}


static inline int cnn_ae_predict(
    const float raw[CNN_WL][CNN_NF],
    float *mse,
    float *max_ratio)
{
    float x[CNN_WL * CNN_NF];

    float encoder[
        CNN_WL * CNN_F1
    ];

    float pooled[
        CNN_POOL_LEN * CNN_F1
    ];

    float latent[
        CNN_POOL_LEN * CNN_F2
    ];

    float upsampled[
        CNN_WL * CNN_F2
    ];

    float decoder[
        CNN_WL * CNN_F1
    ];

    float reconstruction[
        CNN_WL * CNN_NF
    ];

    float feature_error[
        CNN_NF
    ] = {0};

    for (int t = 0; t < CNN_WL; t++)
    {
        for (int f = 0; f < CNN_NF; f++)
        {
            float physical =
                raw[t][f] / CNN_DIV;

            float range =
                CNN_MAX[f] - CNN_MIN[f];

            x[t * CNN_NF + f] =
                range > 0.0f
                ? (
                    physical - CNN_MIN[f]
                  ) / range
                : 0.0f;
        }
    }

    cnn_conv_same(
        x,
        CNN_WL,
        CNN_NF,
        CNN_K1,
        CNN_B1,
        CNN_F1,
        1,
        encoder
    );

    cnn_pool_2(
        encoder,
        CNN_F1,
        pooled
    );

    cnn_conv_same(
        pooled,
        CNN_POOL_LEN,
        CNN_F1,
        CNN_K2,
        CNN_B2,
        CNN_F2,
        1,
        latent
    );

    cnn_upsample_2(
        latent,
        CNN_F2,
        upsampled
    );

    cnn_conv_same(
        upsampled,
        CNN_WL,
        CNN_F2,
        CNN_K3,
        CNN_B3,
        CNN_F1,
        1,
        decoder
    );

    cnn_conv_same(
        decoder,
        CNN_WL,
        CNN_F1,
        CNN_K4,
        CNN_B4,
        CNN_NF,
        0,
        reconstruction
    );

    float total_error = 0.0f;

    for (int t = 0; t < CNN_WL; t++)
    {
        for (int f = 0; f < CNN_NF; f++)
        {
            int index =
                t * CNN_NF + f;

            float difference =
                x[index]
                - reconstruction[index];

            float squared_error =
                difference * difference;

            total_error += squared_error;
            feature_error[f] += squared_error;
        }
    }

    *mse =
        total_error
        / (CNN_WL * CNN_NF);

    *max_ratio = 0.0f;

    for (int f = 0; f < CNN_NF; f++)
    {
        float mean_feature_error =
            feature_error[f] / CNN_WL;

        float ratio =
            mean_feature_error
            / fmaxf(
                CNN_FT[f],
                1.0e-12f
            );

        if (ratio > *max_ratio)
        {
            *max_ratio = ratio;
        }
    }

    return (
        *mse > CNN_RT
    )
    ||
    (
        *max_ratio > CNN_MRT
    );
}

#endif
''')

print(
    "Saved:",
    C_HEADER_PATH,
)

print(
    "Header size KB:",
    round(
        os.path.getsize(
            C_HEADER_PATH
        ) / 1024,
        2,
    ),
)
```

```python
# ============================================================
# CELL 18 GENERATE cnn_autoencoder_params.h FOR STM32 X-CUBE-AI
# Run after final_scaler and final thresholds are calculated
# ============================================================

from pathlib import Path
import numpy as np

# ------------------------------------------------------------
# Deployment configuration
# ------------------------------------------------------------

CNN_AE_N_FEATURES = 14
CNN_AE_WINDOW_LEN = 30
CNN_AE_WINDOW_STEP = 10

# Use 1.0 because STM32 supplies physical float values.
CNN_AE_SCALE_DIVISOR = 1.0

HEADER_FILE = Path("cnn_autoencoder_params.h")

FEATURE_NAMES = [
    "T_slope_10s",
    "PWM",
    "Irms",
    "Vrms",
    "Current_Variance",
    "Current_Crest_Factor",
    "Vib_RMS",
    "Vib_Min",
    "Vib_Q1",
    "Vib_Median",
    "Vib_Q3",
    "Vib_Max",
    "Vib_Variance",
    "Vib_Skewness",
]

# ------------------------------------------------------------
# Verify that all required final-training variables exist
# ------------------------------------------------------------

required_variables = [
    "final_scaler",
    "FINAL_RECON_THRESHOLD",
    "FINAL_MAX_RATIO_THRESHOLD",
    "FINAL_FEATURE_THRESHOLDS",
]

missing_variables = [
    variable_name
    for variable_name in required_variables
    if variable_name not in globals()
]

if missing_variables:
    raise RuntimeError(
        "The following variables are missing:\n"
        + "\n".join(f"  - {name}" for name in missing_variables)
        + "\n\nRun this cell only after final training, scaling and "
          "threshold calculation."
    )

# ------------------------------------------------------------
# Read the exact values from the final training run
# ------------------------------------------------------------

cnn_min = np.asarray(
    final_scaler.data_min_,
    dtype=np.float64
).reshape(-1)

cnn_max = np.asarray(
    final_scaler.data_max_,
    dtype=np.float64
).reshape(-1)

feature_thresholds = np.asarray(
    FINAL_FEATURE_THRESHOLDS,
    dtype=np.float64
).reshape(-1)

global_recon_threshold = float(FINAL_RECON_THRESHOLD)
max_ratio_threshold = float(FINAL_MAX_RATIO_THRESHOLD)

# ------------------------------------------------------------
# Validate dimensions and numeric values
# ------------------------------------------------------------

if len(FEATURE_NAMES) != CNN_AE_N_FEATURES:
    raise ValueError(
        f"FEATURE_NAMES contains {len(FEATURE_NAMES)} features, "
        f"but CNN_AE_N_FEATURES is {CNN_AE_N_FEATURES}."
    )

if cnn_min.size != CNN_AE_N_FEATURES:
    raise ValueError(
        f"Scaler minimum array has {cnn_min.size} values; "
        f"expected {CNN_AE_N_FEATURES}."
    )

if cnn_max.size != CNN_AE_N_FEATURES:
    raise ValueError(
        f"Scaler maximum array has {cnn_max.size} values; "
        f"expected {CNN_AE_N_FEATURES}."
    )

if feature_thresholds.size != CNN_AE_N_FEATURES:
    raise ValueError(
        f"Feature-threshold array has {feature_thresholds.size} values; "
        f"expected {CNN_AE_N_FEATURES}."
    )

if not np.all(np.isfinite(cnn_min)):
    raise ValueError("CNN_AE_MIN contains NaN or infinity.")

if not np.all(np.isfinite(cnn_max)):
    raise ValueError("CNN_AE_MAX contains NaN or infinity.")

if not np.all(np.isfinite(feature_thresholds)):
    raise ValueError(
        "CNN_AE_FEATURE_ERROR_THRESHOLD contains NaN or infinity."
    )

if not np.isfinite(global_recon_threshold):
    raise ValueError("FINAL_RECON_THRESHOLD is not finite.")

if not np.isfinite(max_ratio_threshold):
    raise ValueError("FINAL_MAX_RATIO_THRESHOLD is not finite.")

if np.any(cnn_max <= cnn_min):
    invalid_features = np.where(cnn_max <= cnn_min)[0]

    raise ValueError(
        "Invalid MinMax range for feature indices: "
        + ", ".join(map(str, invalid_features))
    )

# ------------------------------------------------------------
# C formatting functions
# ------------------------------------------------------------

def c_float(value):
    """Convert a Python number to a valid float32 C literal."""
    return f"{float(value):.10e}f"


def create_c_array(array_name, values, comments):
    lines = [
        f"static const float {array_name}[CNN_AE_N_FEATURES] = {{"
    ]

    for index, (value, comment) in enumerate(zip(values, comments)):
        comma = "," if index < len(values) - 1 else ""

        lines.append(
            f"    {c_float(value)}{comma}  "
            f"/* [{index:2d}] {comment} */"
        )

    lines.append("};")

    return "\n".join(lines)

# ------------------------------------------------------------
# Generate header contents
# ------------------------------------------------------------

min_array_text = create_c_array(
    "CNN_AE_MIN",
    cnn_min,
    FEATURE_NAMES,
)

max_array_text = create_c_array(
    "CNN_AE_MAX",
    cnn_max,
    FEATURE_NAMES,
)

feature_threshold_array_text = create_c_array(
    "CNN_AE_FEATURE_ERROR_THRESHOLD",
    feature_thresholds,
    FEATURE_NAMES,
)

header_text = f"""\
#ifndef CNN_AUTOENCODER_PARAMS_H
#define CNN_AUTOENCODER_PARAMS_H

/*
 * Automatically generated from the final CNN autoencoder run.
 *
 * Feature order must be identical on the PC and STM32:
 *
 *  0: T_slope_10s
 *  1: PWM
 *  2: Irms
 *  3: Vrms
 *  4: Current_Variance
 *  5: Current_Crest_Factor
 *  6: Vib_RMS
 *  7: Vib_Min
 *  8: Vib_Q1
 *  9: Vib_Median
 * 10: Vib_Q3
 * 11: Vib_Max
 * 12: Vib_Variance
 * 13: Vib_Skewness
 */

#define CNN_AE_N_FEATURES       {CNN_AE_N_FEATURES}
#define CNN_AE_WINDOW_LEN       {CNN_AE_WINDOW_LEN}
#define CNN_AE_WINDOW_STEP      {CNN_AE_WINDOW_STEP}

#define CNN_AE_INPUT_SIZE \\
    (CNN_AE_WINDOW_LEN * CNN_AE_N_FEATURES)

/*
 * Keep this at 1.0f when the MCU input window contains physical
 * floating-point values.
 */
#define CNN_AE_SCALE_DIVISOR \\
    {c_float(CNN_AE_SCALE_DIVISOR)}

#define CNN_AE_RECON_THRESHOLD \\
    {c_float(global_recon_threshold)}

#define CNN_AE_MAX_RATIO_THRESHOLD \\
    {c_float(max_ratio_threshold)}

{min_array_text}

{max_array_text}

{feature_threshold_array_text}

/*
 * Convert one physical feature value to the MinMax-scaled value
 * expected by the CNN autoencoder.
 *
 * No clipping is applied because clipping would hide values that
 * fall outside the normal training range.
 */
static inline float cnn_ae_scale_raw(
    float raw_value,
    int feature_id)
{{
    float physical;
    float denominator;

    if ((feature_id < 0) ||
        (feature_id >= CNN_AE_N_FEATURES))
    {{
        return 0.0f;
    }}

    physical =
        raw_value / CNN_AE_SCALE_DIVISOR;

    denominator =
        CNN_AE_MAX[feature_id]
        - CNN_AE_MIN[feature_id];

    if (denominator == 0.0f)
    {{
        return 0.0f;
    }}

    return (
        physical - CNN_AE_MIN[feature_id]
    ) / denominator;
}}

/*
 * Calculate the ratio between a feature MSE and its normal
 * feature-specific threshold.
 */
static inline float cnn_ae_feature_error_ratio(
    float feature_mse,
    int feature_id)
{{
    float threshold;

    if ((feature_id < 0) ||
        (feature_id >= CNN_AE_N_FEATURES))
    {{
        return 0.0f;
    }}

    threshold =
        CNN_AE_FEATURE_ERROR_THRESHOLD[feature_id];

    if (threshold <= 0.0f)
    {{
        return 0.0f;
    }}

    return feature_mse / threshold;
}}

/*
 * Final CNN-AE anomaly decision.
 */
static inline int cnn_ae_window_is_anomaly(
    float reconstruction_mse,
    float maximum_feature_ratio)
{{
    return (
        (reconstruction_mse >
            CNN_AE_RECON_THRESHOLD)
        ||
        (maximum_feature_ratio >
            CNN_AE_MAX_RATIO_THRESHOLD)
    );
}}

#endif /* CNN_AUTOENCODER_PARAMS_H */
"""

# ------------------------------------------------------------
# Save and display verification information
# ------------------------------------------------------------

HEADER_FILE.write_text(
    header_text,
    encoding="utf-8",
)

print("=" * 70)
print("CNN-AE PARAMETER HEADER GENERATED")
print("=" * 70)
print(f"File:                  {HEADER_FILE.resolve()}")
print(f"Features:              {CNN_AE_N_FEATURES}")
print(f"Window:                {CNN_AE_WINDOW_LEN} × {CNN_AE_N_FEATURES}")
print(f"Input elements:        {CNN_AE_WINDOW_LEN * CNN_AE_N_FEATURES}")
print(f"Window step:           {CNN_AE_WINDOW_STEP}")
print(f"Scale divisor:         {CNN_AE_SCALE_DIVISOR}")
print(f"Reconstruction limit:  {global_recon_threshold:.10f}")
print(f"Maximum ratio limit:   {max_ratio_threshold:.10f}")
print("=" * 70)

print("\nScaler and feature thresholds:\n")

for index, feature_name in enumerate(FEATURE_NAMES):
    print(
        f"{index:2d}  {feature_name:24s} "
        f"min={cnn_min[index]: .10f}  "
        f"max={cnn_max[index]: .10f}  "
        f"threshold={feature_thresholds[index]: .10f}"
    )

print("\nHeader preview:\n")
print(header_text)

# ------------------------------------------------------------
# Automatically download when running in Google Colab
# ------------------------------------------------------------

try:
    from google.colab import files
    files.download(str(HEADER_FILE))
except ImportError:
    print(
        "\nNot running in Google Colab. "
        "The header was saved in the current notebook folder."
    )
```

```python
# ==========================================================
# CELL 19 — manual test function
# ==========================================================
def manual_1dcnn_autoencoder_prediction(
    test_file,
):
    model = tf.keras.models.load_model(
        KERAS_PATH,
        custom_objects={
            "window_mse_loss":
                window_mse_loss,
        },
    )

    scaler = joblib.load(
        SCALER_PATH
    )

    thresholds = np.load(
        THRESHOLDS_PATH
    )

    reconstruction_threshold = float(
        thresholds[
            "recon_threshold"
        ]
    )

    feature_thresholds = (
        thresholds[
            "feature_thresholds"
        ].astype(
            np.float32
        )
    )

    max_ratio_threshold = float(
        thresholds[
            "max_ratio_threshold"
        ]
    )

    file_fraction_threshold = float(
        thresholds[
            "file_anomaly_fraction_threshold"
        ]
    )

    test_df = read_one(
        test_file
    )

    X_test, starts = make_windows_from_array(
        test_df.to_numpy(
            dtype=np.float32
        ),
        scaler,
    )

    (
        reconstruction_mse,
        per_feature_error,
        worst_feature_ids,
    ) = anomaly_score(
        model,
        X_test,
    )

    (
        flags,
        maximum_feature_ratio,
    ) = decisions_from_errors(
        reconstruction_mse,
        per_feature_error,
        reconstruction_threshold,
        feature_thresholds,
        max_ratio_threshold,
    )

    anomaly_fraction = float(
        np.mean(flags)
    )

    file_decision = (
        "ANOMALY"
        if anomaly_fraction
        >= file_fraction_threshold
        else "HEALTHY"
    )

    result = pd.DataFrame({
        "window": np.arange(
            len(X_test)
        ),

        "start_row":
            starts,

        "time_sec": (
            starts
            + WINDOW_LEN / 2.0
        ) * ROW_PERIOD_SEC,

        "reconstruction_mse":
            reconstruction_mse,

        "reconstruction_threshold":
            reconstruction_threshold,

        "max_feature_ratio":
            maximum_feature_ratio,

        "max_ratio_threshold":
            max_ratio_threshold,

        "worst_feature": [
            FEATURE_COLUMNS[int(index)]
            for index in worst_feature_ids
        ],

        "decision": np.where(
            flags == 1,
            "anomaly",
            "normal",
        ),

        "is_anomaly": flags,
    })

    stem = Path(
        test_file
    ).stem

    output_csv = os.path.join(
        MANUAL_TEST_DIR,
        f"manual_test_{stem}.csv",
    )

    result.to_csv(
        output_csv,
        index=False,
    )

    print("\nWindow decisions:")
    print(result["decision"].value_counts())

    print("\nMost frequent worst features:")
    print(result["worst_feature"].value_counts().head(8))
    print(
        "\nMANUAL LSTM AUTOENCODER TEST"
    )

    print(
        "=" * 60
    )

    print(
        "File:",
        test_file,
    )

    print(
        "Total windows:",
        len(result),
    )
    print("Normal windows:", int(np.sum(flags == 0)))
    
    print(
        "Anomalous windows:",
        int(
            np.sum(flags)
        ),
    )

    print(
        "Anomaly fraction:",
        anomaly_fraction,
    )

    print(
        "File decision:",
        file_decision,
    )

    print(
        "Reconstruction threshold:",
        reconstruction_threshold,
    )

    print(
        "Saved CSV:",
        output_csv,
    )

    plt.figure(
        figsize=(12, 4)
    )

    plt.plot(
        result[
            "time_sec"
        ],
        result[
            "reconstruction_mse"
        ],
    )

    plt.axhline(
        reconstruction_threshold,
        linestyle="--",
        label="reconstruction threshold",
    )

    plt.xlabel(
        "Time [s]"
    )

    plt.ylabel(
        "Reconstruction MSE"
    )

    plt.title(
        f"1D CNN Autoencoder Manual Test — {stem}"
    )

    plt.legend()
    plt.tight_layout()

    plt.savefig(
        os.path.join(
            MANUAL_TEST_DIR,
            f"manual_test_{stem}_mse.png",
        ),
        dpi=150,
    )

    plt.show()

    plt.figure(
        figsize=(12, 4)
    )

    plt.step(
        result[
            "time_sec"
        ],
        result[
            "is_anomaly"
        ],
        where="mid",
    )

    plt.yticks(
        [0, 1],
        [
            "normal",
            "anomaly",
        ],
    )

    plt.xlabel(
        "Time [s]"
    )

    plt.ylabel(
        "Decision"
    )

    plt.title(
        f"1D CNN Autoencoder Decisions — {stem}"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            MANUAL_TEST_DIR,
            f"manual_test_{stem}_timeline.png",
        ),
        dpi=150,
    )

    plt.show()

    return result
```

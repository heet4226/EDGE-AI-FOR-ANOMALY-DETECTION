```python
# ==========================================================
# CELL 1 — imports
# ==========================================================
import os
import glob
import random
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.utils.class_weight import compute_sample_weight

```

```python
# ==========================================================
# CELL 2 — configuration
# ==========================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)

BASE_PATH = (
    r"../../data"
    r"\dataset\Motor_data_model"
)

SCALE_DIVISOR = 1_000_000.0

FEATURE_COLUMNS = [
    "T_slope_10s",
    "PWM",
    "Irms",
    "Vrms",
    "Current_Variance",
    "Current_Crest_Factor",
    "Vib_RMS",
    "Vib_Min",
    "Vib_Q1",
    "Vib_Median",
    "Vib_Q3",
    "Vib_Max",
    "Vib_Variance",
    "Vib_Skewness",
]

FINAL_LABELS = [
    "motor_off",
    "turning_off",
    "turning_on",
    "normal",
    "electrical_fault",
    "mechanical_fault",
    "thermal_fault",
]

LABEL_TO_ID = {
    label: index
    for index, label in enumerate(FINAL_LABELS)
}

N_FEATURES = len(FEATURE_COLUMNS)
N_CLASSES = len(FINAL_LABELS)
WINDOW_LEN = 30
WINDOW_STEP = 10
INPUT_SIZE = WINDOW_LEN * N_FEATURES
ROW_PERIOD_SEC = 0.015

N_SPLITS = 5

OUTPUT_DIR = 'xgboost_output'
CV_DIR = os.path.join(OUTPUT_DIR, "cross_validation")
FINAL_DIR = os.path.join(OUTPUT_DIR, "final_model")
MANUAL_TEST_DIR = os.path.join(OUTPUT_DIR, "manual_test_results")

for folder in [CV_DIR, FINAL_DIR, MANUAL_TEST_DIR]:
    os.makedirs(folder, exist_ok=True)

print("Model:", 'XGBoost Classifier')
print("Input:", WINDOW_LEN, "x", N_FEATURES)
print("Classes:", FINAL_LABELS)

```

```python
# ==========================================================
# CELL 3 — map each file to one class
# ==========================================================
def normalized_path(path):
    return str(path).replace("\\", "/").lower()


def label_for(path):
    path_text = normalized_path(path)
    stem = Path(path).stem.lower()

    if (
        "motor_being_off" in path_text
        or "turning_off" in path_text
        or "transition_off" in path_text
        or "off_spikes" in path_text
    ):
        return "turning_off"

    if (
        "motor_being_on" in path_text
        or "turning_on" in path_text
        or "transition_on" in path_text
        or "on_spikes" in path_text
    ):
        return "turning_on"

    if "/motor_off/" in path_text or "motor_off" in stem:
        return "motor_off"

    if (
        "electrical fault" in path_text
        or "electrical_fault" in path_text
        or "abnormal_current" in path_text
    ):
        return "electrical_fault"

    if (
        "mechanical fault" in path_text
        or "mechanical_fault" in path_text
        or "abnormal_vibration" in path_text
    ):
        return "mechanical_fault"

    if (
        "thermal fault" in path_text
        or "thermal_fault" in path_text
        or "abnormal_temp" in path_text
    ):
        return "thermal_fault"

    if (
        "/normal/" in path_text
        or stem.startswith("normal_")
        or stem == "normal"
    ):
        return "normal"

    return None
```

```python
# ==========================================================
# CELL 4 — discover and read files
# ==========================================================
def discover_files(base_path):
    files = glob.glob(
        os.path.join(base_path, "**", "*.xlsx"),
        recursive=True,
    )
    files += glob.glob(
        os.path.join(base_path, "**", "*.csv"),
        recursive=True,
    )
    return sorted(set(files))


def read_one(path):
    path = Path(path)

    if path.suffix.lower() == ".xlsx":
        df = pd.read_excel(path)
    else:
        df = pd.read_csv(path)

    missing_columns = [
        column
        for column in FEATURE_COLUMNS
        if column not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Missing columns in {path.name}: {missing_columns}"
        )

    clean = df[FEATURE_COLUMNS].apply(
        pd.to_numeric,
        errors="coerce",
    )

    clean = clean / SCALE_DIVISOR
    clean = clean.replace([np.inf, -np.inf], np.nan)
    clean = clean.dropna().reset_index(drop=True)

    if len(clean) < WINDOW_LEN:
        raise ValueError(
            f"{path.name} has only {len(clean)} valid rows."
        )

    return clean
```

```python
# ==========================================================
# CELL 5 — load labeled sessions
# ==========================================================
files = discover_files(BASE_PATH)

if not files:
    raise RuntimeError(
        f"No CSV or Excel files found under: {BASE_PATH}"
    )

sessions = []

for file_path in files:
    label = label_for(file_path)

    if label is None:
        print(
            "SKIP unrecognized:",
            os.path.relpath(file_path, BASE_PATH),
        )
        continue

    try:
        df = read_one(file_path)
    except Exception as error:
        print(
            "SKIP read error:",
            os.path.relpath(file_path, BASE_PATH),
            "|",
            error,
        )
        continue

    sessions.append({
        "X": df.to_numpy(dtype=np.float32),
        "file": os.path.relpath(file_path, BASE_PATH),
        "label": label,
        "class_id": LABEL_TO_ID[label],
        "rows": len(df),
    })

if not sessions:
    raise RuntimeError("No labeled sessions were loaded.")

class_counts = pd.Series([
    session["label"]
    for session in sessions
]).value_counts().reindex(
    FINAL_LABELS,
    fill_value=0,
)

print("\nFILES PER CLASS")
print(class_counts)

missing_classes = [
    label
    for label in FINAL_LABELS
    if class_counts[label] == 0
]

if missing_classes:
    raise RuntimeError(
        f"No files found for classes: {missing_classes}"
    )
```

```python
# ==========================================================
# CELL 6 — create flattened 30 × 14 windows
# ==========================================================
def make_windows_from_array(X_raw):
    starts = np.arange(
        0,
        len(X_raw) - WINDOW_LEN + 1,
        WINDOW_STEP,
        dtype=np.int64,
    )

    if len(starts) == 0:
        return None, None

    windows = np.stack([
        X_raw[start:start + WINDOW_LEN]
        for start in starts
    ]).astype(np.float32)

    flattened = windows.reshape(
        len(windows),
        INPUT_SIZE,
    )

    return flattened, starts


def build_dataset_from_sessions(selected_sessions):
    all_windows = []
    all_labels = []
    all_files = []
    all_starts = []

    for session in selected_sessions:
        windows, starts = make_windows_from_array(
            session["X"]
        )

        if windows is None:
            continue

        all_windows.append(windows)
        all_labels.extend(
            [session["class_id"]] * len(windows)
        )
        all_files.extend(
            [session["file"]] * len(windows)
        )
        all_starts.extend(starts.tolist())

    if not all_windows:
        raise RuntimeError("No windows were created.")

    return (
        np.concatenate(all_windows, axis=0),
        np.asarray(all_labels, dtype=np.int64),
        np.asarray(all_files, dtype=object),
        np.asarray(all_starts, dtype=np.int64),
    )
```

```python
# ==========================================================
# CELL 7 — flattened feature names
# ==========================================================
FLAT_FEATURE_NAMES = []

for time_index in range(WINDOW_LEN):
    for feature_name in FEATURE_COLUMNS:
        FLAT_FEATURE_NAMES.append(
            f"t{time_index:02d}_{feature_name}"
        )
```

```python
# ==========================================================
# CELL 8 — build classifier
# ==========================================================
from xgboost import XGBClassifier


def build_classifier():
    return XGBClassifier(
        n_estimators=20,
        max_depth=3,
        learning_rate=0.10,
        min_child_weight=2,
        subsample=0.90,
        colsample_bytree=0.80,
        gamma=0.0,
        reg_alpha=0.0,
        reg_lambda=1.0,
        objective="multi:softprob",
        num_class=7,
        eval_metric="mlogloss",
        tree_method="hist",
        random_state=42,
        n_jobs=-1,
    )


temporary_model = build_classifier()
print(temporary_model.get_params())
del temporary_model
```

```python
# ==========================================================
# CELL 9 — five-fold file-wise cross-validation
# ==========================================================
cv_table = pd.DataFrame({
    "label": [
        session["class_id"]
        for session in sessions
    ],
    "group": [
        session["file"]
        for session in sessions
    ],
})

splitter = StratifiedGroupKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=SEED,
)

cv_metrics = []
all_true = []
all_pred = []
all_window_predictions = []

for fold, (train_index, test_index) in enumerate(
    splitter.split(
        cv_table,
        cv_table["label"],
        groups=cv_table["group"],
    ),
    start=1,
):
    print("\n" + "=" * 60)
    print("FOLD", fold)
    print("=" * 60)

    train_sessions = [
        sessions[index]
        for index in train_index
    ]

    test_sessions = [
        sessions[index]
        for index in test_index
    ]

    X_train, y_train, _, _ = (
        build_dataset_from_sessions(
            train_sessions
        )
    )

    (
        X_test,
        y_test,
        test_files,
        test_starts,
    ) = build_dataset_from_sessions(
        test_sessions
    )

    sample_weight = compute_sample_weight(
        class_weight="balanced",
        y=y_train,
    )

    model = build_classifier()

    model.fit(
        X_train,
        y_train,
        sample_weight=sample_weight,
    )

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)
    confidence = np.max(probabilities, axis=1)

    all_true.extend(y_test.tolist())
    all_pred.extend(predictions.tolist())

    fold_metrics = {
        "fold": fold,
        "accuracy": accuracy_score(
            y_test,
            predictions,
        ),
        "balanced_accuracy": balanced_accuracy_score(
            y_test,
            predictions,
        ),
        "macro_precision": precision_score(
            y_test,
            predictions,
            average="macro",
            zero_division=0,
        ),
        "macro_recall": recall_score(
            y_test,
            predictions,
            average="macro",
            zero_division=0,
        ),
        "macro_f1": f1_score(
            y_test,
            predictions,
            average="macro",
            zero_division=0,
        ),
    }

    cv_metrics.append(fold_metrics)

    fold_predictions = pd.DataFrame({
        "fold": fold,
        "file": test_files,
        "start_row": test_starts,
        "true_class_id": y_test,
        "true_label": [
            FINAL_LABELS[index]
            for index in y_test
        ],
        "pred_class_id": predictions,
        "pred_label": [
            FINAL_LABELS[index]
            for index in predictions
        ],
        "confidence": confidence,
    })

    all_window_predictions.append(
        fold_predictions
    )

    print(pd.Series(fold_metrics))

```

```python
# ==========================================================
# CELL 10 — combined cross-validation report
# ==========================================================
cv_metrics_df = pd.DataFrame(cv_metrics)

cv_predictions_df = pd.concat(
    all_window_predictions,
    ignore_index=True,
)

cv_metrics_df.to_csv(
    os.path.join(CV_DIR, "cv_metrics.csv"),
    index=False,
)

cv_predictions_df.to_csv(
    os.path.join(CV_DIR, "cv_window_predictions.csv"),
    index=False,
)

all_true = np.asarray(
    all_true,
    dtype=np.int64,
)

all_pred = np.asarray(
    all_pred,
    dtype=np.int64,
)

overall_metrics = {
    "accuracy": accuracy_score(
        all_true,
        all_pred,
    ),
    "balanced_accuracy": balanced_accuracy_score(
        all_true,
        all_pred,
    ),
    "macro_precision": precision_score(
        all_true,
        all_pred,
        average="macro",
        zero_division=0,
    ),
    "macro_recall": recall_score(
        all_true,
        all_pred,
        average="macro",
        zero_division=0,
    ),
    "macro_f1": f1_score(
        all_true,
        all_pred,
        average="macro",
        zero_division=0,
    ),
}

print("\nOVERALL CROSS-VALIDATION RESULTS")
print(pd.Series(overall_metrics))

print(
    classification_report(
        all_true,
        all_pred,
        labels=np.arange(N_CLASSES),
        target_names=FINAL_LABELS,
        zero_division=0,
    )
)

pd.DataFrame([
    overall_metrics
]).to_csv(
    os.path.join(CV_DIR, "overall_metrics.csv"),
    index=False,
)

ConfusionMatrixDisplay(
    confusion_matrix(
        all_true,
        all_pred,
        labels=np.arange(N_CLASSES),
    ),
    display_labels=FINAL_LABELS,
).plot(
    values_format="d",
    xticks_rotation=45,
)

plt.title(
    "XGBoost Classifier Cross-Validation"
)

plt.tight_layout()

plt.savefig(
    os.path.join(CV_DIR, "confusion_matrix.png"),
    dpi=180,
)

plt.show()

```

```python
# ==========================================================
# CELL 11 — train final model on all sessions
# ==========================================================
X_all, y_all, _, _ = (
    build_dataset_from_sessions(
        sessions
    )
)

final_sample_weight = compute_sample_weight(
    class_weight="balanced",
    y=y_all,
)

final_model = build_classifier()

final_model.fit(
    X_all,
    y_all,
    sample_weight=final_sample_weight,
)
```

```python
# ==========================================================
# CELL 12 — save model, labels, and metadata
# ==========================================================
MODEL_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "xgboost_classifier.pkl",
    )
)

LABELS_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "xgboost_classifier_labels.txt",
    )
)

METADATA_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "xgboost_classifier_metadata.csv",
    )
)

os.makedirs(
    FINAL_DIR,
    exist_ok=True,
)

joblib.dump(
    final_model,
    MODEL_PATH,
)

with open(
    LABELS_PATH,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "\n".join(FINAL_LABELS)
    )

pd.DataFrame({
    "flat_feature_id": np.arange(INPUT_SIZE),
    "flat_feature_name": FLAT_FEATURE_NAMES,
}).to_csv(
    METADATA_PATH,
    index=False,
)

print("Saved model:", MODEL_PATH)
print("Model exists:", os.path.exists(MODEL_PATH))
print("Model size KB:", round(os.path.getsize(MODEL_PATH) / 1024, 2))
print("Saved labels:", LABELS_PATH)
print("Saved metadata:", METADATA_PATH)
```

```python
# ==========================================================
# CELL 13 — feature importance report
# ==========================================================
importance_df = pd.DataFrame({
    "flat_feature_name": FLAT_FEATURE_NAMES,
    "importance": final_model.feature_importances_,
})

importance_df = importance_df.sort_values(
    "importance",
    ascending=False,
).reset_index(drop=True)

importance_df.to_csv(
    os.path.join(
        FINAL_DIR,
        "feature_importance.csv",
    ),
    index=False,
)

print("\nTOP 30 FEATURES")
print(importance_df.head(30))
```

```python
# ==========================================================
# CELL 14 — save model-specific native files
# ==========================================================
JSON_MODEL_PATH = os.path.join(
    FINAL_DIR,
    "xgboost_classifier.json",
)

UBJ_MODEL_PATH = os.path.join(
    FINAL_DIR,
    "xgboost_classifier.ubj",
)

final_model.save_model(JSON_MODEL_PATH)
final_model.save_model(UBJ_MODEL_PATH)

print("Saved:", JSON_MODEL_PATH)
print("Saved:", UBJ_MODEL_PATH)

print(
    "JSON model size KB:",
    round(
        os.path.getsize(JSON_MODEL_PATH) / 1024,
        2,
    ),
)

print(
    "UBJ model size KB:",
    round(
        os.path.getsize(UBJ_MODEL_PATH) / 1024,
        2,
    ),
)

```

```python
# ==========================================================
# CELL 15 — export ONNX
# ==========================================================
from onnxmltools import convert_xgboost
from onnxmltools.convert.common.data_types import FloatTensorType

ONNX_PATH = os.path.join(
    FINAL_DIR,
    "xgboost_classifier.onnx",
)

onnx_model = convert_xgboost(
    final_model,
    initial_types=[
        (
            "float_input",
            FloatTensorType([
                None,
                INPUT_SIZE,
            ]),
        )
    ],
    target_opset=13,
)

with open(
    ONNX_PATH,
    "wb",
) as file:
    file.write(
        onnx_model.SerializeToString()
    )

print("Saved:", ONNX_PATH)
```

```python
# ==========================================================
# CELL 15.1 — export parameter-only header (for ONNX/pkl runtime, class labels + dims)
# ==========================================================
PARAMS_HEADER_PATH = os.path.join(
    FINAL_DIR,
    "xgboost_params.h",
)

with open(PARAMS_HEADER_PATH, "w", encoding="utf-8") as file:
    file.write("#ifndef XGBOOST_PARAMS_H\n")
    file.write("#define XGBOOST_PARAMS_H\n\n")

    file.write(f"#define XGB_WINDOW_LEN    {WINDOW_LEN}\n")
    file.write(f"#define XGB_N_FEATURES    {N_FEATURES}\n")
    file.write(f"#define XGB_INPUT_SIZE    {INPUT_SIZE}\n")
    file.write(f"#define XGB_N_CLASSES     {N_CLASSES}\n")
    file.write(f"#define XGB_SCALE_DIVISOR {SCALE_DIVISOR:.9g}f\n\n")

    file.write("/* Class labels, in model output order (index = predict_proba argmax) */\n")
    file.write("static const char *XGB_LABELS[XGB_N_CLASSES] = {\n")
    for label in FINAL_LABELS:
        file.write(f'    "{label}",\n')
    file.write("};\n\n")

    file.write("#endif /* XGBOOST_PARAMS_H */\n")

print("Saved:", PARAMS_HEADER_PATH)
```

```python
# ==========================================================
# CELL 16 — manual classification test
# ==========================================================
def manual_classification_prediction(
    test_file,
):
    model_path = os.path.abspath(
        os.path.join(
            "xgboost_output",
            "final_model",
            "xgboost_classifier.pkl",
        )
    )

    print("Loading model from:", model_path)

    if not os.path.exists(
        model_path
    ):
        raise FileNotFoundError(
            f"Model file not found:\n"
            f"{model_path}\n\n"
            "Run Cell 11 and Cell 12 first."
        )

    model = joblib.load(
        model_path
    )

    test_df = read_one(
        test_file
    )

    X_test, starts = make_windows_from_array(
        test_df.to_numpy(
            dtype=np.float32
        )
    )

    if (
        X_test is None
        or len(X_test) == 0
    ):
        raise RuntimeError(
            "The selected file does not contain enough "
            "valid rows for one 30-row window."
        )

    predictions = model.predict(
        X_test
    )

    probabilities = model.predict_proba(
        X_test
    )

    confidence = np.max(
        probabilities,
        axis=1,
    )

    result = pd.DataFrame({
        "window": np.arange(
            len(X_test)
        ),
        "start_row": starts,
        "time_sec": (
            starts + WINDOW_LEN / 2.0
        ) * ROW_PERIOD_SEC,
        "pred_class_id": predictions,
        "pred_label": [
            FINAL_LABELS[
                int(index)
            ]
            for index in predictions
        ],
        "confidence": confidence,
    })

    for class_id, label in enumerate(
        FINAL_LABELS
    ):
        result[
            f"probability_{label}"
        ] = probabilities[
            :,
            class_id,
        ]

    majority_class_id = int(
        result[
            "pred_class_id"
        ].value_counts().idxmax()
    )

    majority_label = FINAL_LABELS[
        majority_class_id
    ]

    stem = Path(
        test_file
    ).stem

    manual_test_dir = os.path.abspath(
        os.path.join(
            "xgboost_output",
            "manual_test_results",
        )
    )

    os.makedirs(
        manual_test_dir,
        exist_ok=True,
    )

    output_csv = os.path.join(
        manual_test_dir,
        f"manual_test_{stem}.csv",
    )

    result.to_csv(
        output_csv,
        index=False,
    )

    print("\n" + "=" * 60)
    print("MANUAL XGBOOST TEST")
    print("=" * 60)
    print("File:", test_file)
    print("Total windows:", len(result))
    print(
        "Majority prediction:",
        majority_label,
    )
    print(
        "Mean confidence:",
        float(
            np.mean(
                confidence
            )
        ),
    )
    print("Saved CSV:", output_csv)

    print("\nWindow predictions:")
    print(
        result[
            "pred_label"
        ].value_counts()
    )

    # ======================================================
    # Manual-file prediction timeline
    # ======================================================
    plt.figure(
        figsize=(12, 4)
    )

    plt.step(
        result["time_sec"],
        result["pred_class_id"],
        where="mid",
    )

    plt.yticks(
        np.arange(
            N_CLASSES
        ),
        FINAL_LABELS,
    )

    plt.xlabel("Time [s]")
    plt.ylabel("Predicted class")

    plt.title(
        f"XGBoost — {stem}"
    )

    plt.tight_layout()

    timeline_path = os.path.join(
        manual_test_dir,
        f"manual_test_{stem}_timeline.png",
    )

    plt.savefig(
        timeline_path,
        dpi=150,
    )

    plt.show()

    print(
        "Saved timeline:",
        timeline_path,
    )

    return result
```

```python
# ==========================================================
# CELL 17 — select manual test file
# ==========================================================
if __name__ == "__main__":
    import tkinter as tk
    from tkinter import filedialog

    root = tk.Tk()
    root.withdraw()
    root.attributes("-topmost", True)

    TEST_FILE = filedialog.askopenfilename(
        title="Select motor CSV or Excel file",
        filetypes=[
            ("Excel files", "*.xlsx"),
            ("CSV files", "*.csv"),
            ("All files", "*.*"),
        ],
    )

    root.destroy()

    if TEST_FILE:
        manual_result = manual_classification_prediction(
            TEST_FILE
        )
    else:
        print("No file selected.")
```

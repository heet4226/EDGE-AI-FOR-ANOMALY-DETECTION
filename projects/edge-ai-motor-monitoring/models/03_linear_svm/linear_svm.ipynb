# Linear SVM Classifier

7-class classifier using a linear-kernel SVM — cheapest possible SVM variant to deploy (no stored support vectors, just a weight vector per class).

```python

# ==========================================================
# CELL 1 — imports
# ==========================================================
import os
import glob
import random
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC

```

```python
# ==========================================================
# CELL 2 — Linear SVM configuration
# ==========================================================
SEED = 42

random.seed(SEED)
np.random.seed(SEED)

BASE_PATH = (
    r"../../data"
    r"/Motor_data_model"
)

SCALE_DIVISOR = 1_000_000.0

FEATURE_COLUMNS = [
    "T_slope_10s",
    "PWM",
    "Irms",
    "Vrms",
    "Current_Variance",
    "Current_Crest_Factor",
    "Vib_RMS",
    "Vib_Min",
    "Vib_Q1",
    "Vib_Median",
    "Vib_Q3",
    "Vib_Max",
    "Vib_Variance",
    "Vib_Skewness",
]

FINAL_LABELS = [
    "motor_off",
    "turning_off",
    "turning_on",
    "normal",
    "electrical_fault",
    "mechanical_fault",
    "thermal_fault",
]

LABEL_TO_ID = {
    label: index
    for index, label in enumerate(FINAL_LABELS)
}

N_FEATURES = len(FEATURE_COLUMNS)
N_CLASSES = len(FINAL_LABELS)

WINDOW_LEN = 30
WINDOW_STEP = 10
INPUT_SIZE = WINDOW_LEN * N_FEATURES
ROW_PERIOD_SEC = 0.015

N_SPLITS = 5

OUTPUT_DIR = "linear_svm_output"
CV_DIR = os.path.join(
    OUTPUT_DIR,
    "cross_validation",
)
FINAL_DIR = os.path.join(
    OUTPUT_DIR,
    "final_model",
)
MANUAL_TEST_DIR = os.path.join(
    OUTPUT_DIR,
    "manual_test_results",
)

for folder in [
    CV_DIR,
    FINAL_DIR,
    MANUAL_TEST_DIR,
]:
    os.makedirs(
        folder,
        exist_ok=True,
    )

print("Model: Linear SVM Classifier")
print("Input:", WINDOW_LEN, "x", N_FEATURES)
print("Flattened input size:", INPUT_SIZE)
print("Classes:", FINAL_LABELS)

```

```python
# ==========================================================
# CELL 3 — map each file path to one supervised class
# ==========================================================
def normalized_path(path):
    return str(path).replace(
        "\\",
        "/",
    ).lower()


def label_for(path):
    path_text = normalized_path(path)
    stem = Path(path).stem.lower()

    if (
        "motor_being_off" in path_text
        or "turning_off" in path_text
        or "transition_off" in path_text
        or "off_spikes" in path_text
    ):
        return "turning_off"

    if (
        "motor_being_on" in path_text
        or "turning_on" in path_text
        or "transition_on" in path_text
        or "on_spikes" in path_text
    ):
        return "turning_on"

    if (
        "/motor_off/" in path_text
        or "motor_off" in stem
    ):
        return "motor_off"

    if (
        "electrical fault" in path_text
        or "electrical_fault" in path_text
        or "abnormal_current" in path_text
    ):
        return "electrical_fault"

    if (
        "mechanical fault" in path_text
        or "mechanical_fault" in path_text
        or "abnormal_vibration" in path_text
    ):
        return "mechanical_fault"

    if (
        "thermal fault" in path_text
        or "thermal_fault" in path_text
        or "abnormal_temp" in path_text
    ):
        return "thermal_fault"

    if (
        "/normal/" in path_text
        or stem.startswith("normal_")
        or stem == "normal"
    ):
        return "normal"

    return None

```

```python
# ==========================================================
# CELL 4 — discover and read CSV / Excel files
# ==========================================================
def discover_files(base_path):
    files = glob.glob(
        os.path.join(
            base_path,
            "**",
            "*.xlsx",
        ),
        recursive=True,
    )

    files += glob.glob(
        os.path.join(
            base_path,
            "**",
            "*.csv",
        ),
        recursive=True,
    )

    return sorted(
        set(files)
    )


def read_one(path):
    path = Path(path)

    if path.suffix.lower() == ".xlsx":
        df = pd.read_excel(path)
    else:
        df = pd.read_csv(path)

    missing_columns = [
        column
        for column in FEATURE_COLUMNS
        if column not in df.columns
    ]

    if missing_columns:
        raise ValueError(
            f"Missing columns in {path.name}: "
            f"{missing_columns}"
        )

    clean = df[
        FEATURE_COLUMNS
    ].apply(
        pd.to_numeric,
        errors="coerce",
    )

    clean = clean / SCALE_DIVISOR

    clean = clean.replace(
        [np.inf, -np.inf],
        np.nan,
    )

    clean = clean.dropna(
    ).reset_index(
        drop=True
    )

    if len(clean) < WINDOW_LEN:
        raise ValueError(
            f"{path.name} has only "
            f"{len(clean)} valid rows."
        )

    return clean
```

```python

# ==========================================================
# CELL 5 — load labeled sessions
# ==========================================================
files = discover_files(
    BASE_PATH
)

if not files:
    raise RuntimeError(
        f"No CSV or Excel files found under: "
        f"{BASE_PATH}"
    )

sessions = []

for file_path in files:
    label = label_for(
        file_path
    )

    if label is None:
        print(
            "SKIP unrecognized:",
            os.path.relpath(
                file_path,
                BASE_PATH,
            ),
        )
        continue

    try:
        df = read_one(
            file_path
        )
    except Exception as error:
        print(
            "SKIP read error:",
            os.path.relpath(
                file_path,
                BASE_PATH,
            ),
            "|",
            error,
        )
        continue

    sessions.append({
        "X": df.to_numpy(
            dtype=np.float32
        ),
        "file": os.path.relpath(
            file_path,
            BASE_PATH,
        ),
        "label": label,
        "class_id": LABEL_TO_ID[
            label
        ],
        "rows": len(df),
    })

if not sessions:
    raise RuntimeError(
        "No labeled sessions were loaded."
    )

class_counts = pd.Series([
    session["label"]
    for session in sessions
]).value_counts().reindex(
    FINAL_LABELS,
    fill_value=0,
)

print("\nFILES PER CLASS")
print(class_counts)

missing_classes = [
    label
    for label in FINAL_LABELS
    if class_counts[label] == 0
]

if missing_classes:
    raise RuntimeError(
        f"No files found for classes: "
        f"{missing_classes}"
    )
```

```python
# ==========================================================
# CELL 6 — create flattened 30 × 14 windows
# ==========================================================
def make_windows_from_array(
    X_raw,
):
    starts = np.arange(
        0,
        len(X_raw) - WINDOW_LEN + 1,
        WINDOW_STEP,
        dtype=np.int64,
    )

    if len(starts) == 0:
        return None, None

    windows = np.stack([
        X_raw[
            start:start + WINDOW_LEN
        ]
        for start in starts
    ]).astype(
        np.float32
    )

    flattened = windows.reshape(
        len(windows),
        INPUT_SIZE,
    )

    return (
        flattened,
        starts,
    )


def build_dataset_from_sessions(
    selected_sessions,
):
    all_windows = []
    all_labels = []
    all_files = []
    all_starts = []

    for session in selected_sessions:
        windows, starts = make_windows_from_array(
            session["X"]
        )

        if windows is None:
            continue

        all_windows.append(
            windows
        )

        all_labels.extend(
            [session["class_id"]]
            * len(windows)
        )

        all_files.extend(
            [session["file"]]
            * len(windows)
        )

        all_starts.extend(
            starts.tolist()
        )

    if not all_windows:
        raise RuntimeError(
            "No windows were created."
        )

    return (
        np.concatenate(
            all_windows,
            axis=0,
        ),
        np.asarray(
            all_labels,
            dtype=np.int64,
        ),
        np.asarray(
            all_files,
            dtype=object,
        ),
        np.asarray(
            all_starts,
            dtype=np.int64,
        ),
    )

```

```python
# ==========================================================
# CELL 7 — flattened feature names
# ==========================================================
FLAT_FEATURE_NAMES = []

for time_index in range(
    WINDOW_LEN
):
    for feature_name in FEATURE_COLUMNS:
        FLAT_FEATURE_NAMES.append(
            f"t{time_index:02d}_"
            f"{feature_name}"
        )
```

```python
# ==========================================================
# CELL 8 — build Linear SVM classifier
# ==========================================================
def build_classifier():
    return Pipeline([
        (
            "scaler",
            StandardScaler(),
        ),
        (
            "classifier",
            LinearSVC(
                penalty="l2",
                loss="squared_hinge",
                dual="auto",
                C=1.0,
                class_weight="balanced",
                max_iter=10000,
                random_state=42,
            ),
        ),
    ])


temporary_model = build_classifier()

print(
    "Linear SVM configuration:"
)

print(
    temporary_model.get_params()
)

del temporary_model

```

```python
# ==========================================================
# CELL 9 — convert decision scores to confidence values
# ==========================================================
def scores_to_probabilities(
    decision_scores,
):
    decision_scores = np.asarray(
        decision_scores,
        dtype=np.float64,
    )

    if decision_scores.ndim == 1:
        decision_scores = np.column_stack([
            -decision_scores,
            decision_scores,
        ])

    shifted_scores = (
        decision_scores
        - np.max(
            decision_scores,
            axis=1,
            keepdims=True,
        )
    )

    exponentials = np.exp(
        shifted_scores
    )

    probabilities = (
        exponentials
        / np.sum(
            exponentials,
            axis=1,
            keepdims=True,
        )
    )

    return probabilities
```

```python
# ==========================================================
# CELL 10 — five-fold file-wise cross-validation
# ==========================================================
cv_table = pd.DataFrame({
    "label": [
        session["class_id"]
        for session in sessions
    ],
    "group": [
        session["file"]
        for session in sessions
    ],
})

splitter = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42,
)

cv_metrics = []
all_true = []
all_pred = []
all_window_predictions = []

for fold, (
    train_index,
    test_index,
) in enumerate(
    splitter.split(
        cv_table,
        cv_table["label"],
        groups=cv_table["group"],
    ),
    start=1,
):
    print("\n" + "=" * 60)
    print("FOLD", fold)
    print("=" * 60)

    train_sessions = [
        sessions[index]
        for index in train_index
    ]

    test_sessions = [
        sessions[index]
        for index in test_index
    ]

    X_train, y_train, _, _ = (
        build_dataset_from_sessions(
            train_sessions
        )
    )

    (
        X_test,
        y_test,
        test_files,
        test_starts,
    ) = build_dataset_from_sessions(
        test_sessions
    )

    model = build_classifier()

    model.fit(
        X_train,
        y_train,
    )

    predictions = model.predict(
        X_test
    )

    decision_scores = model.decision_function(
        X_test
    )

    probabilities = scores_to_probabilities(
        decision_scores
    )

    confidence = np.max(
        probabilities,
        axis=1,
    )

    all_true.extend(
        y_test.tolist()
    )

    all_pred.extend(
        predictions.tolist()
    )

    fold_metrics = {
        "fold": fold,
        "accuracy": accuracy_score(
            y_test,
            predictions,
        ),
        "balanced_accuracy":
            balanced_accuracy_score(
                y_test,
                predictions,
            ),
        "macro_precision":
            precision_score(
                y_test,
                predictions,
                average="macro",
                zero_division=0,
            ),
        "macro_recall":
            recall_score(
                y_test,
                predictions,
                average="macro",
                zero_division=0,
            ),
        "macro_f1":
            f1_score(
                y_test,
                predictions,
                average="macro",
                zero_division=0,
            ),
    }

    cv_metrics.append(
        fold_metrics
    )

    fold_predictions = pd.DataFrame({
        "fold": fold,
        "file": test_files,
        "start_row": test_starts,
        "true_class_id": y_test,
        "true_label": [
            FINAL_LABELS[index]
            for index in y_test
        ],
        "pred_class_id": predictions,
        "pred_label": [
            FINAL_LABELS[index]
            for index in predictions
        ],
        "confidence": confidence,
    })

    all_window_predictions.append(
        fold_predictions
    )

    print(
        pd.Series(
            fold_metrics
        )
    )

```

```python
# ==========================================================
# CELL 11 — combined cross-validation report
# ==========================================================
cv_metrics_df = pd.DataFrame(
    cv_metrics
)

cv_predictions_df = pd.concat(
    all_window_predictions,
    ignore_index=True,
)

cv_metrics_df.to_csv(
    os.path.join(
        CV_DIR,
        "cv_metrics.csv",
    ),
    index=False,
)

cv_predictions_df.to_csv(
    os.path.join(
        CV_DIR,
        "cv_window_predictions.csv",
    ),
    index=False,
)

all_true = np.asarray(
    all_true,
    dtype=np.int64,
)

all_pred = np.asarray(
    all_pred,
    dtype=np.int64,
)

overall_metrics = {
    "accuracy": accuracy_score(
        all_true,
        all_pred,
    ),
    "balanced_accuracy":
        balanced_accuracy_score(
            all_true,
            all_pred,
        ),
    "macro_precision":
        precision_score(
            all_true,
            all_pred,
            average="macro",
            zero_division=0,
        ),
    "macro_recall":
        recall_score(
            all_true,
            all_pred,
            average="macro",
            zero_division=0,
        ),
    "macro_f1":
        f1_score(
            all_true,
            all_pred,
            average="macro",
            zero_division=0,
        ),
}

print(
    "\nOVERALL CROSS-VALIDATION RESULTS"
)

print(
    pd.Series(
        overall_metrics
    )
)

print(
    classification_report(
        all_true,
        all_pred,
        labels=np.arange(
            N_CLASSES
        ),
        target_names=FINAL_LABELS,
        zero_division=0,
    )
)

pd.DataFrame([
    overall_metrics
]).to_csv(
    os.path.join(
        CV_DIR,
        "overall_metrics.csv",
    ),
    index=False,
)

ConfusionMatrixDisplay(
    confusion_matrix(
        all_true,
        all_pred,
        labels=np.arange(
            N_CLASSES
        ),
    ),
    display_labels=FINAL_LABELS,
).plot(
    values_format="d",
    xticks_rotation=45,
)

plt.title(
    "Linear SVM Classifier Cross-Validation"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        CV_DIR,
        "confusion_matrix.png",
    ),
    dpi=180,
)

plt.show()

```

```python
# ==========================================================
# CELL 12 — train final Linear SVM model
# ==========================================================
X_all, y_all, _, _ = (
    build_dataset_from_sessions(
        sessions
    )
)

final_model = Pipeline([
    (
        "scaler",
        StandardScaler(),
    ),
    (
        "classifier",
        LinearSVC(
            penalty="l2",
            loss="squared_hinge",
            dual="auto",
            C=1.0,
            class_weight="balanced",
            max_iter=10000,
            random_state=42,
        ),
    ),
])

final_model.fit(
    X_all,
    y_all,
)

print(
    "Final model trained on windows:",
    len(X_all),
)

```

```python
# ==========================================================
# CELL 13 — save model, labels, and metadata
# ==========================================================
MODEL_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "linear_svm_classifier.pkl",
    )
)

LABELS_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "linear_svm_labels.txt",
    )
)

METADATA_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "linear_svm_metadata.csv",
    )
)

joblib.dump(
    final_model,
    MODEL_PATH,
)

with open(
    LABELS_PATH,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "\n".join(
            FINAL_LABELS
        )
    )

pd.DataFrame({
    "flat_feature_id": np.arange(
        INPUT_SIZE
    ),
    "flat_feature_name":
        FLAT_FEATURE_NAMES,
}).to_csv(
    METADATA_PATH,
    index=False,
)

print("Saved model:", MODEL_PATH)

print(
    "Model size KB:",
    round(
        os.path.getsize(
            MODEL_PATH
        ) / 1024,
        2,
    ),
)

print("Saved labels:", LABELS_PATH)
print("Saved metadata:", METADATA_PATH)

```

```python
# ==========================================================
# CELL 14 — coefficient importance report
# ==========================================================
classifier = final_model.named_steps[
    "classifier"
]

coefficient_matrix = classifier.coef_

mean_absolute_importance = np.mean(
    np.abs(
        coefficient_matrix
    ),
    axis=0,
)

importance_df = pd.DataFrame({
    "flat_feature_name":
        FLAT_FEATURE_NAMES,
    "mean_absolute_coefficient":
        mean_absolute_importance,
})

importance_df = importance_df.sort_values(
    "mean_absolute_coefficient",
    ascending=False,
).reset_index(
    drop=True
)

importance_df.to_csv(
    os.path.join(
        FINAL_DIR,
        "coefficient_importance.csv",
    ),
    index=False,
)

print(
    "\nTOP 30 LINEAR SVM FEATURES"
)

print(
    importance_df.head(30)
)
```

```python
# ==========================================================
# CELL 15 — export complete Linear SVM pipeline to ONNX
# ==========================================================
from skl2onnx import convert_sklearn
from skl2onnx.common.data_types import FloatTensorType

ONNX_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "linear_svm_model.onnx",
    )
)

trained_classifier = final_model.named_steps[
    "classifier"
]

onnx_model = convert_sklearn(
    final_model,
    initial_types=[
        (
            "float_input",
            FloatTensorType([
                None,
                INPUT_SIZE,
            ]),
        )
    ],
    options={
        id(trained_classifier): {
            "raw_scores": True,
        }
    },
    target_opset=13,
)

with open(
    ONNX_PATH,
    "wb",
) as file:
    file.write(
        onnx_model.SerializeToString()
    )

print("Saved complete ONNX:", ONNX_PATH)

print(
    "ONNX size KB:",
    round(
        os.path.getsize(
            ONNX_PATH
        ) / 1024,
        2,
    ),
)

print(
    "This ONNX contains both StandardScaler "
    "and LinearSVC."
)
```

```python
# ==========================================================
# CELL 15A — generate/download Linear SVM parameter header
# ==========================================================
# Generate and download only the parameter header

# Generate and download only the parameter header
from pathlib import Path

HEADER_PATH = Path(FINAL_DIR) / "linear_svm_params.h"

with open(HEADER_PATH, "w", encoding="utf-8") as file:
    file.write("""#ifndef LINEAR_SVM_PARAMS_H
#define LINEAR_SVM_PARAMS_H

#define LSVM_WINDOW_LEN   30U
#define LSVM_N_FEATURES   14U
#define LSVM_INPUT_SIZE   420U
#define LSVM_N_CLASSES    7U

/*
 * StandardScaler is already included inside the ONNX model.
 * Supply live physical FLOAT32 values directly.
 */
#define LSVM_EXTERNAL_SCALING  0U
#define LSVM_LIVE_DIVISION     0U

static const char *const LSVM_LABELS[LSVM_N_CLASSES] = {
    "motor_off",
    "turning_off",
    "turning_on",
    "normal",
    "electrical_fault",
    "mechanical_fault",
    "thermal_fault"
};

#endif /* LINEAR_SVM_PARAMS_H */
""")

print("Created:", HEADER_PATH)
```

```python
# ==========================================================
# CELL 16 — export direct C header
# ==========================================================
HEADER_PATH = os.path.abspath(
    os.path.join(
        FINAL_DIR,
        "linear_svm_classifier_model.h",
    )
)

scaler = final_model.named_steps[
    "scaler"
]

classifier = final_model.named_steps[
    "classifier"
]

scaler_mean = scaler.mean_.astype(
    np.float32
)

scaler_scale = scaler.scale_.astype(
    np.float32
)

weights = classifier.coef_.astype(
    np.float32
)

bias = classifier.intercept_.astype(
    np.float32
)


def write_float_array(
    file,
    name,
    values,
):
    values = np.asarray(
        values,
        dtype=np.float32,
    ).reshape(-1)

    file.write(
        f"static const float {name}"
        f"[{len(values)}] = {{\n"
    )

    for start in range(
        0,
        len(values),
        8,
    ):
        line = ", ".join(
            f"{float(value):.9g}f"
            for value in values[
                start:start + 8
            ]
        )

        file.write(
            f"    {line},\n"
        )

    file.write(
        "};\n\n"
    )


with open(
    HEADER_PATH,
    "w",
    encoding="utf-8",
) as file:
    file.write(
        "#ifndef LINEAR_SVM_CLASSIFIER_MODEL_H\n"
    )

    file.write(
        "#define LINEAR_SVM_CLASSIFIER_MODEL_H\n\n"
    )

    file.write(
        "#include <math.h>\n\n"
    )

    file.write(
        f"#define LSVM_WL {WINDOW_LEN}\n"
    )

    file.write(
        f"#define LSVM_NF {N_FEATURES}\n"
    )

    file.write(
        f"#define LSVM_INPUT {INPUT_SIZE}\n"
    )

    file.write(
        f"#define LSVM_CLASSES {N_CLASSES}\n"
    )

    file.write(
        f"#define LSVM_DIV {SCALE_DIVISOR:.9g}f\n\n"
    )

    write_float_array(
        file,
        "LSVM_MEAN",
        scaler_mean,
    )

    write_float_array(
        file,
        "LSVM_SCALE",
        scaler_scale,
    )

    write_float_array(
        file,
        "LSVM_WEIGHT",
        weights,
    )

    write_float_array(
        file,
        "LSVM_BIAS",
        bias,
    )

    c_source = r"""
static inline int linear_svm_predict(
    const float raw[LSVM_WL][LSVM_NF],
    float probability[LSVM_CLASSES],
    float *confidence)
{
    float scores[LSVM_CLASSES];

    for (int class_index = 0;
         class_index < LSVM_CLASSES;
         class_index++)
    {
        float sum = LSVM_BIAS[class_index];

        for (int time_index = 0;
             time_index < LSVM_WL;
             time_index++)
        {
            for (int feature_index = 0;
                 feature_index < LSVM_NF;
                 feature_index++)
            {
                int flat_index =
                    time_index * LSVM_NF
                    + feature_index;

                float physical =
                    raw[time_index][feature_index]
                    / LSVM_DIV;

                float scaled =
                    LSVM_SCALE[flat_index] > 0.0f
                    ? (
                        physical
                        - LSVM_MEAN[flat_index]
                    ) / LSVM_SCALE[flat_index]
                    : 0.0f;

                sum += scaled
                    * LSVM_WEIGHT[
                        class_index * LSVM_INPUT
                        + flat_index
                    ];
            }
        }

        scores[class_index] = sum;
    }

    int predicted_class = 0;
    float maximum_score = scores[0];

    for (int class_index = 1;
         class_index < LSVM_CLASSES;
         class_index++)
    {
        if (
            scores[class_index]
            > maximum_score
        )
        {
            maximum_score =
                scores[class_index];

            predicted_class =
                class_index;
        }
    }

    float denominator = 0.0f;

    for (int class_index = 0;
         class_index < LSVM_CLASSES;
         class_index++)
    {
        probability[class_index] =
            expf(
                scores[class_index]
                - maximum_score
            );

        denominator +=
            probability[class_index];
    }

    float maximum_probability = 0.0f;

    for (int class_index = 0;
         class_index < LSVM_CLASSES;
         class_index++)
    {
        probability[class_index] /=
            denominator;

        if (
            probability[class_index]
            > maximum_probability
        )
        {
            maximum_probability =
                probability[class_index];
        }
    }

    *confidence =
        maximum_probability;

    return predicted_class;
}

#endif
"""

    file.write(
        c_source
    )

print("Saved:", HEADER_PATH)

print(
    "Header size KB:",
    round(
        os.path.getsize(
            HEADER_PATH
        ) / 1024,
        2,
    ),
)

```

```python
# ==========================================================
# CELL 17 — verify direct C-style prediction
# ==========================================================
def direct_python_prediction(
    flattened_window,
):
    scaled = (
        flattened_window
        - scaler_mean
    ) / scaler_scale

    scores = (
        weights @ scaled
        + bias
    )

    return int(
        np.argmax(
            scores
        )
    )


number_check = min(
    100,
    len(X_all),
)

sklearn_predictions = final_model.predict(
    X_all[:number_check]
)

direct_predictions = []

for window in X_all[
    :number_check
]:
    class_id = direct_python_prediction(
        window
    )

    direct_predictions.append(
        class_id
    )

direct_predictions = np.asarray(
    direct_predictions,
    dtype=np.int64,
)

if not np.array_equal(
    sklearn_predictions,
    direct_predictions,
):
    raise RuntimeError(
        "Direct Linear SVM calculation "
        "does not match sklearn."
    )

print(
    "Direct C-style Linear SVM verified on",
    number_check,
    "windows.",
)
```

```python
# ==========================================================
# CELL 18 — manual classification test
# ==========================================================
def manual_classification_prediction(
    test_file,
):
    model_path = os.path.abspath(
        os.path.join(
            "linear_svm_output",
            "final_model",
            "linear_svm_classifier.pkl",
        )
    )

    print(
        "Loading model from:",
        model_path,
    )

    if not os.path.exists(
        model_path
    ):
        raise FileNotFoundError(
            f"Model file not found:\n"
            f"{model_path}\n\n"
            "Run Cell 12 and Cell 13 first."
        )

    model = joblib.load(
        model_path
    )

    test_df = read_one(
        test_file
    )

    X_test, starts = make_windows_from_array(
        test_df.to_numpy(
            dtype=np.float32
        )
    )

    if (
        X_test is None
        or len(X_test) == 0
    ):
        raise RuntimeError(
            "The selected file does not contain enough "
            "valid rows for one 30-row window."
        )

    predictions = model.predict(
        X_test
    )

    decision_scores = model.decision_function(
        X_test
    )

    probabilities = scores_to_probabilities(
        decision_scores
    )

    confidence = np.max(
        probabilities,
        axis=1,
    )

    result = pd.DataFrame({
        "window": np.arange(
            len(X_test)
        ),
        "start_row": starts,
        "time_sec": (
            starts + WINDOW_LEN / 2.0
        ) * ROW_PERIOD_SEC,
        "pred_class_id":
            predictions,
        "pred_label": [
            FINAL_LABELS[
                int(index)
            ]
            for index in predictions
        ],
        "confidence":
            confidence,
    })

    for class_id, label in enumerate(
        FINAL_LABELS
    ):
        result[
            f"score_{label}"
        ] = decision_scores[
            :,
            class_id,
        ]

        result[
            f"probability_{label}"
        ] = probabilities[
            :,
            class_id,
        ]

    majority_class_id = int(
        result[
            "pred_class_id"
        ].value_counts(
        ).idxmax()
    )

    majority_label = FINAL_LABELS[
        majority_class_id
    ]

    stem = Path(
        test_file
    ).stem

    manual_test_dir = os.path.abspath(
        os.path.join(
            "linear_svm_output",
            "manual_test_results",
        )
    )

    os.makedirs(
        manual_test_dir,
        exist_ok=True,
    )

    output_csv = os.path.join(
        manual_test_dir,
        f"manual_test_{stem}.csv",
    )

    result.to_csv(
        output_csv,
        index=False,
    )

    print("\n" + "=" * 60)
    print("MANUAL LINEAR SVM TEST")
    print("=" * 60)
    print("File:", test_file)
    print("Total windows:", len(result))
    print(
        "Majority prediction:",
        majority_label,
    )
    print(
        "Mean confidence:",
        float(
            np.mean(
                confidence
            )
        ),
    )
    print("Saved CSV:", output_csv)

    print(
        "\nWindow predictions:"
    )

    print(
        result[
            "pred_label"
        ].value_counts()
    )

    # ======================================================
    # Manual-file prediction timeline
    # ======================================================
    plt.figure(
        figsize=(12, 4)
    )

    plt.step(
        result["time_sec"],
        result["pred_class_id"],
        where="mid",
    )

    plt.yticks(
        np.arange(
            N_CLASSES
        ),
        FINAL_LABELS,
    )

    plt.xlabel("Time [s]")
    plt.ylabel("Predicted class")

    plt.title(
        f"Linear SVM — {stem}"
    )

    plt.tight_layout()

    timeline_path = os.path.join(
        manual_test_dir,
        f"manual_test_{stem}_timeline.png",
    )

    plt.savefig(
        timeline_path,
        dpi=150,
    )

    plt.show()

    print(
        "Saved timeline:",
        timeline_path,
    )

    return result

```
